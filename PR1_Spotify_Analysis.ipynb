## Задание 1. Загрузка данных
Подключаем основные библиотеки для работы с данными и визуализации (Pandas, NumPy, Matplotlib) и загружаем CSV-файл с музыкальными треками в DataFrame.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Загружаем CSV-файл в DataFrame с именем df
df = pd.read_csv('spotify_artist_streaming_2020_2025.csv')
print("Файл успешно загружен!")
```

*Комментарий к коду:* Код импортирует необходимые библиотеки `pandas`, `numpy`, `matplotlib.plt` и загружает данные из файла `spotify_artist_streaming_2020_2025.csv` в переменную `df`.

## Задание 2. Первичное знакомство с датасетом
Выполняем вывод первых строк таблицы, определяем количество строк и столбцов, выводим названия столбцов и проверяем типы данных.

```python
# Вывод первых 10 строк таблицы
print("--- Первые 10 строк датасета ---")
display(df.head(10))

# Определение количества строк и столбцов
print(f"Размер таблицы (строки, столбцы): {df.shape}")

# Вывод названий всех столбцов
print("\n--- Названия столбцов ---")
print(df.columns.tolist())

# Информация о типах данных столбцов
print("\n--- Информация о типах данных ---")
df.info()
```

## Задание 3. Сортировка данных
Находим 10 треков с максимальным количеством прослушиваний (`stream_count`).

```python
top10_streamed = df.sort_values(by='stream_count', ascending=False).head(10)
display(top10_streamed[['artist_name', 'track_name', 'genre', 'stream_count']])
```

*Комментарий к коду:* Функция `sort_values` отсортировала датасет по убыванию значений прослушиваний.

**Ответ на вопрос:** Среди самых прослушиваемых треков встречаются исполнители: *Grove Nova, Flora Quinn, Ethereal, SethStone, IsleEyre, Dean System, Rocco Ore, Knot Hart, The Inks, Prism Hart*.  
Жанры лидирующих треков: *Alternative, Rock, Pop, Metal, Hip-Hop, R&B*.

## Задание 4. Фильтрация
Находим треки, выпущенные после 2024 года (`release_year > 2024`).

```python
tracks_after_2024 = df[df['release_year'] > 2024]
print(f"Количество треков, выпущенных после 2024 года: {len(tracks_after_2024)}")
display(tracks_after_2024.head())
```

*Комментарий к коду:* С помощью фильтрации `df['release_year'] > 2024` мы отобрали только треки 2025 года выпуска.

## Задание 5. Основные статистические показатели
Находим статистику для показателя популярности (`popularity`).

```python
pop_mean = df['popularity'].mean()
pop_median = df['popularity'].median()
pop_min = df['popularity'].min()
pop_max = df['popularity'].max()
pop_std = df['popularity'].std()

print(f"Среднее значение (Mean): {pop_mean:.2f}")
print(f"Медиана (Median): {pop_median:.2f}")
print(f"Минимум (Min): {pop_min}")
print(f"Максимум (Max): {pop_max}")
print(f"Стандартное отклонение (Std): {pop_std:.2f}")

print("\n--- Сводная статистика функцией describe() ---")
display(df['popularity'].describe())
```

*Пояснение к статистическим показателям:*
- **Mean (Среднее)** = 28.14 — среднее арифметическое всех значений популярности.
- **Median (Медиана)** = 26.00 — серединное значение, разграничивающее верхнюю и нижнюю половины выборки.
- **Min (Минимум)** = 0 — минимальное зарегистрированное значение популярности.
- **Max (Максимум)** = 92 — максимальное значение популярности трека.
- **Std (Стандартное отклонение)** = 16.00 — показывает степень разброса значений относительно среднего значения.

## Задание 6. Мода
Определяем самые часто встречающиеся значения (моду) для столбцов `genre` и `release_year`.

```python
genre_mode = df['genre'].mode()[0]
year_mode = df['release_year'].mode()[0]

print(f"Мода для столбца genre (самый частый жанр): {genre_mode}")
print(f"Мода для столбца release_year (самый частый год выпуска): {year_mode}")
```

*Комментарий к коду:* Функция `.mode()` вернет наиболее часто встречающиеся категории. Самым частым жанром является **R&B**, а самым частым годом выпуска — **2024**.

## Задание 7. Гистограмма
Строим гистограмму распределения показателей популярности (`popularity`).

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=30, color='skyblue', edgecolor='black')
plt.title('Распределение популярности треков (Popularity)', fontsize=14)
plt.xlabel('Популярность (Popularity)', fontsize=12)
plt.ylabel('Количество треков', fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

**Ответы на вопросы по графику:**
1. **В каком диапазоне находится основная часть значений?** Основная масса треков расположена в диапазоне популярности от **10 до 45** (с пиком в районе 20-30).
2. **Есть ли значения, которые встречаются значительно реже остальных?** Да, значения популярности выше **60-70** встречаются значительно реже остальных, являясь редкими высокорейтинговыми треками.
