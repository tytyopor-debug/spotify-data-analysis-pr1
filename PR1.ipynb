## Задание 1. Загрузка данных

```python
# Подключаем библиотеки для анализа данных и визуализации
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Считываем датасет из CSV-файла в датафрейм df
df = pd.read_csv('spotify_artist_streaming_2020_2025.csv')
```

Импортировали библиотеки и считали файл в dataframe df.

## Задание 2. Первичное знакомство с датасетом

```python
# Смотрим первые 10 строк таблицы для первичной оценки
display(df.head(10))

# Выводим размеры таблицы (количество строк и столбцов)
print("Размер таблицы:", df.shape)

# Получаем полный список названий всех колонок
print("Столбцы:", df.columns)

# Выводим общую информацию: типы данных и наличие пропусков
df.info()
```

Посмотрели первые 10 строк, размер таблицы (50000 строк и 33 столбца), названия колонок и типы данных.

## Задание 3. Сортировка данных

```python
# Сортируем датасет по убыванию количества прослушиваний и берем первые 10
top10 = df.sort_values(by='stream_count', ascending=False).head(10)

# Выводим только ключевые столбцы для топ-10 треков
top10[['artist_name', 'track_name', 'genre', 'stream_count']]
```

Отсортировали данные по убыванию stream_count.  
Среди самых прослушиваемых треков встречаются исполнители Grove Nova, Flora Quinn, Ethereal, SethStone, IsleEyre и др. Жанры: Alternative, Rock, Pop, Metal, Hip-Hop, R&B.

## Задание 4. Фильтрация

```python
# Фильтруем записи: оставляем только треки, выпущенные строго после 2024 года
df[df['release_year'] > 2024].head()
```

Отфильтровали треки, выпущенные после 2024 года.

## Задание 5. Основные статистические показатели

```python
# Считаем отдельные метрики по популярности поочередно
print("Среднее:", df['popularity'].mean())
print("Медиана:", df['popularity'].median())
print("Минимум:", df['popularity'].min())
print("Максимум:", df['popularity'].max())
print("Стандартное отклонение:", df['popularity'].std())

# Выводим сводную статистику одной функцией .describe()
df['popularity'].describe()
```

Посчитали статистику по столбцу popularity: среднее — 28.14, медиана — 26, мин — 0, макс — 92, стд — 16. Команда describe покажет все эти метрики сразу.

## Задание 6. Мода

```python
# Находим самое часто повторяющееся значение (моду) для жанров и года выпуска
print("Мода genre:", df['genre'].mode()[0])
print("Мода release_year:", df['release_year'].mode()[0])
```

Нашли самые частые значения: для genre это R&B, для release_year — 2024.

## Задание 7. Гистограмма

```python
# Строим гистограмму распределения популярности (разбиваем на 20 интервалов)
plt.hist(df['popularity'], bins=20, edgecolor='black')

# Настраиваем название и подписи осей графика
plt.title('Распределение popularity')
plt.xlabel('Popularity')
plt.ylabel('Количество')
plt.show()
```

1. Основная часть значений популярности находится в диапазоне от 10 до 45.
2. Значения выше 60 встречаются значительно реже остальных.
